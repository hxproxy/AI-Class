# Question
Is there an association between gender and reaction time among teenagers?

## Study
I will be conducting an observational study due to me only having a sample for students in America and that I'm only looking at students in highschool (9-12) because they are less likely to put joke answers.

```python
import matplotlib.pyplot as plt
import seaborn as sns
import pandas as pd
import pathlib
```

```python
# had to REALLY think to get this to work (i remembered lambda is a thing yay)
df = pd.read_csv(pathlib.Path('../../data/CensusAtSchool_Sample.csv'))
clean_df = df.dropna(subset=['Gender', 'Reaction_time']).copy()
clean_df['Reaction_time'] = pd.to_numeric(clean_df['Reaction_time'], errors='coerce')
clean_df = clean_df.dropna(subset=['Reaction_time'])
clean_df = clean_df[(clean_df['Reaction_time'] > 0) & (clean_df['Reaction_time'] < 5)]
clean_df['Speed_Category'] = clean_df['Reaction_time'].apply(
    lambda x: 'Fast (<= 0.5s)' if x <= 0.5 else 'Slow (> 0.5s)'
)
```

```python
sns.set_theme('paper')
sns.set_style('darkgrid')
plt.figure(figsize=(8, 5))

sns.countplot(
    data=clean_df, 
    x="Speed_Category", 
    hue="Gender", 
    palette=["#ec83e7", "#66abfa"]
)

plt.title("Reaction Time Category by Gender")
plt.xlabel("Speed Category")
plt.ylabel("Count")

plt.tight_layout()
plt.show()

pd.crosstab(clean_df, ['Gender', 'Reaction_time'])

print(clean_df[['Gender', 'Reaction_time']])
```
