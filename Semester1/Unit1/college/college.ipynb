# Question
What are the graduation rates for White vs Asian races, at M.I.T and Cal Tech?

## Study
I will conduct an observational study because I am only looking at data for 2 schools.

```python
import matplotlib.pyplot as plt
import seaborn as sns
import pandas as pd
import pathlib
```

```python
df = pd.read_csv(pathlib.Path('../../data/cleaned_college.csv'))
s = df[df['school_name'] == "Massachusetts Institute of Technology"]
s2 = df[df['school_name'] == "California Institute of Technology"]
white_mit = s.grad_rate_white
asian_mit = s.grad_rate_asian
white_cal = s2.grad_rate_white
asian_cal = s2.grad_rate_asian
```

```python
sns.set_theme('paper')
sns.set_style('darkgrid')
sns.set_palette(["#ff9a2a"])

def get_zscore(data: float | int, mean: float | int, std: float | int) -> float | int:
  return (data - mean) / std

white_mit_zscore = float(f"{get_zscore(df['grad_rate_white'].std(), white_mit.mean(), white_mit.item()):.3f}")
asian_mit_zscore = float(f"{get_zscore(df['grad_rate_asian'].std(), asian_mit.mean(), asian_mit.item()):.3f}")
white_cal_zscore = float(f"{get_zscore(df['grad_rate_white'].std(), white_cal.mean(), white_cal.item()):.3f}")
asian_cal_zscore = float(f"{get_zscore(df['grad_rate_asian'].std(), asian_cal.mean(), asian_cal.item()):.3f}")

z_score_data = pd.DataFrame({
    'Group': ['White - MIT', 'Asian - MIT', 'White - Cal', 'Asian - Cal'],
    'School': ['MIT', 'MIT', 'Cal Tech', 'Cal Tech'],
    'Ethnicity': ['White', 'Asian', 'White', 'Asian'],
    'Z_Score': [white_mit_zscore, asian_mit_zscore, white_cal_zscore, asian_cal_zscore]
})

#print(f"z w m: {white_mit_zscore}\nz a m: {asian_mit_zscore}\nz w c: {white_cal_zscore}\nz a c: {asian_mit_zscore}")

def showplot():
  sns.kdeplot(z_score_data, x="Z_Score").set(xlabel='Z-Score')

  plt.axvline(white_mit_zscore, color="#ff4545", linestyle="--", label="MIT")
  plt.text(
    white_mit_zscore, 0.95,
    f"MIT White",
    transform=plt.gca().get_xaxis_transform(),
    ha="left",
    va="top"
  )
  plt.axvline(asian_mit_zscore, color="#ff4545", linestyle="--")
  plt.text(
    asian_mit_zscore, 0.95,
    f"MIT Asian",
    transform=plt.gca().get_xaxis_transform(),
    ha="left",
    va="top"
  )
  plt.axvline(white_cal_zscore, color="#45caff", linestyle="--", label="Cal Tech")
  plt.text(
    white_cal_zscore, 0.95,
    f"Cal White",
    transform=plt.gca().get_xaxis_transform(),
    ha="left",
    va="bottom"
  )
  plt.axvline(asian_cal_zscore, color="#45caff", linestyle="--")
  plt.text(
    asian_cal_zscore, 0.95,
    f"Cal Asian",
    transform=plt.gca().get_xaxis_transform(),
    ha="left",
    va="bottom"
  )
  plt.legend(title="School", loc="upper left")
  plt.show()
```

# Conclusion
In conclusion, Asian People are most likely to graduate from M.I.T Or Cal Tech. This is is proven because for Asian People at M.I.T have a Z-Score of **-0.705** while White People have a Z-Score of **-0.764** and Asian People have a Z-Score of **-0.705** at Cal Tech while White People have a Z-Score of **-0.760**.<br>
