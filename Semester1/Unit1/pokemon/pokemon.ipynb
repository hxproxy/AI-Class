# Question
how does Eevee's defense compare to all it's evoloutions?

## Study
I will conduct a census study to answer this question because I have access to data for every evoloution of Eevee, which is a population dataset. I will not be including shiny or any other varients of the eveloutions.

```python
import matplotlib.pyplot as plt
import seaborn as sns
import pandas as pd
import pathlib
```

```python
df = pd.read_csv(pathlib.Path('../../data/pokemon.csv')) # finally no more full paths cause i remembered pathlib exists
whitelist = ["vaporeon","jolteon","flareon","leafeon","glaceon","sylveon","espeon","umbreon"]
```

```python
sns.set_theme('paper')
sns.set_style('darkgrid')

data = []

for i in range(len(df)):
  name = str(df.english_name[i]).lower()

  if name in whitelist or name == "eevee":
    data.append({
      "name": str(df.english_name[i]).lower(),
      "defense": int(df.defense[i])
    })

data = pd.DataFrame(data)
defense = data.defense

def showplt():
  sns.histplot(data, x="defense", color="#ff9a2a")
  plt.axvline(defense.median(), color="#ff4545", linestyle="--")
  plt.text(
    defense.median(), 0.95,
    f"Median: {defense.median():.0f}",
    transform=plt.gca().get_xaxis_transform(),
    ha="left",
    va="top"
  )
  plt.axvline(defense.min(), color="#45f3ff", linestyle="--")
  plt.text(
    defense.min(), 0.95,
    f"Min: {defense.min():.0f}",
    transform=plt.gca().get_xaxis_transform(),
    ha="left",
    va="top"
  )
  plt.axvline(defense.max(), color="#35db07", linestyle="--")
  plt.text(
    defense.max(), 0.95,
    f"Max: {defense.max():.0f}",
    transform=plt.gca().get_xaxis_transform(),
    ha="left",
    va="top"
  )
  plt.axvline(defense.mean(), color="#d807db", linestyle="--")
  plt.text(
    defense.mean(), 0.95,
    f"Mean: {defense.mean():.2f}",
    transform=plt.gca().get_xaxis_transform(),
    ha="left",
    va="top"
  )
  plt.show()

#print(data)

# removed all output so it looks pretty
```

# Conclusion
In conclusion the evolutions of Eevee are stronger than it when it comes to defense. This is shown in the data because Eevee has a defense of `50` while the strongest evolution *(Leafeon)* has a defense of `130` and in general a median of `60`.
