### My Question
What is the probability of rolling a six on a single die out of 10,000?

### My Study
Describe your simulation here:
1. What is the **random process** being simulated?<br>
Rolling a die 10,000 times to see how many times it lands on 6<br>
2. What will one **trial** look like? What are the possible **outcomes** of this trial?<br>
1 trial will be a dataframe of 10,000 rolls and then I will calculate the relative frequency<br>
3. What **event** are you interested in? Which outcomes are included in that event?<br>
I am interested to see how the data varies<br>

```python
import matplotlib.pyplot as plt
import seaborn as sns
import pandas as pd
import numpy as np

sns.set_theme('paper')
sns.set_style('darkgrid')
alternating_colors = ["#ff9a2a", "#931ce3"] * 3

rng = np.random.default_rng()

num_rolls = 10000
rolls = rng.integers(1, 7, size=num_rolls)
df_rolls = pd.DataFrame({
  "roll_number": np.arange(1, num_rolls + 1),
  "outcome": rolls
})

count_sixes = (rolls == 6).sum()

relative_frequency = count_sixes / 10000

#print(f"Number of times a 6 was rolled: {count_sixes}")
#print(f"Relative frequency: {relative_frequency}")
#print(f"Theoretical probability (1/6): {1/6:.4f}")

def showgraph():
  plt.figure(figsize=(8, 4))
  sns.countplot(data=df_rolls, x="outcome", palette=alternating_colors, hue="outcome", legend=False)

  plt.axhline(10000 / 6, color="red", linestyle="--", label="Expected (~1667)")

  plt.title("Frequency of Each Die Face in 10,000 Rolls")
  plt.xlabel("Die Face")
  plt.ylabel("Count")
  plt.legend()
  plt.show()
```

### My Conclusion
In this simulation of rolling a die 10,000 times, a 6 was rolled 1,670 times, yielding an empirical probability (relative frequency) of 0.167. This experimental result aligns almost perfectly with the theoretical probability of 1/6≈0.1667.

This experiment demonstrates the Law of Large Numbers: as the number of trials increases, the empirical probability converges toward the theoretical probability, proving that the simulated die behaves like a fair six-sided die.
