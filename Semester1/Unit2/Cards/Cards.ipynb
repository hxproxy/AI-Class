# Question
What is the probability that you pull a skip turn card from a deck and it being blue in UNO?

### Simulation
I will simulate pulling a card from a deck of Uno cards. A trial will look like "pulling" a card from a list in a variable 1,500 times and the result is appended to another variable which records the ammount of times a card was pulled.

```python
import matplotlib.pyplot as plt
from collections import Counter
import seaborn as sns
import random as rnd
import pandas as pd
```

```python
cards = ['red0','red1','red2','red3','red4','red5','red6','red7','red8','red9','red_skip','red_reverse','red_+2','wild','yellow0','yellow1','yellow2','yellow3','yellow4','yellow5','yellow6','yellow7','yellow8','yellow9','yellow_skip','yellow_reverse','yellow_+2','wild','green0','green1','green2','green3','green4','green5','green6','green7','green8','green9','green_skip','green_reverse','green_+2','wild','blue0','blue1','blue2','blue3','blue4','blue5','blue6','blue7','blue8','blue9','blue_skip','blue_reverse','blue_+2','wild','red1','red2','red3','red4','red5','red6','red7','red8','red9','red_skip','red_reverse','red_+2','+4','yellow1','yellow2','yellow3','yellow4','yellow5','yellow6','yellow7','yellow8','yellow9','yellow_skip','yellow_reverse','yellow_+2','+4','green1','green2','green3','green4','green5','green6','green7','green8','green9','green_skip','green_reverse','green_+2','+4','blue1','blue2','blue3','blue4','blue5','blue6','blue7','blue8','blue9','blue_skip','blue_reverse','blue_+2','+4']
num_draws = 1500
```

```python
drawn_cards = [rnd.choice(cards) for _ in range(num_draws)] 
card_counts = Counter(drawn_cards)
df_counts = pd.DataFrame(
  list(card_counts.items()), columns=["Card", "Times Pulled"]
).sort_values(by="Times Pulled", ascending=False)
```

```python
blue_skip_count = df_counts.loc[df_counts['Card'] == 'blue_skip', 'Times Pulled'].values

bcount = blue_skip_count[0] if len(blue_skip_count) > 0 else 0
#print(f"blue_skip was pulled {bcount} time(s).")
```

```python
def showgraph():
  sns.set_theme('paper')
  sns.set_style('darkgrid')
  sns.set_palette(["#ff9a2a"])

  plt.figure(figsize=(12, 5))

  # Pass the raw drawn_cards list directly
  sns.countplot(x=drawn_cards, order=df_counts["Card"])

  plt.xticks(rotation=45, ha="right")
  plt.xlabel("Card")
  plt.ylabel("Times Pulled")
  plt.title("Card Pull Frequencies (50 Draws)")
  plt.tight_layout()

  plt.show()
```

# Conclusion
Based on the simulation of pulling a card from an UNO deck 1,500 times, a blue skip card was drawn 23 times. This gives an experimental probability of approximately 1.53% (or 23/1500). Theoretically, a standard 108-card UNO deck contains 2 blue skip cards, yielding a theoretical probability of about 1.85% (2/108).<br>
