# Question
How many turns before you expect someone to land on somebody's property in Monopoly? Assume there are properties on every 3 spaces.

### Simulation
I will simulate taking turns in a Monopoly game that is about halfway through so that there are properties that are owned by then.

```python
import matplotlib.pyplot as plt
from tkinter import filedialog
import seaborn as sns
import tkinter as tk
import numpy as np
```

```python
BOARD_SIZE = 40
NUM_PLAYERS = 4
PROPERTIES = list(range(0, BOARD_SIZE, 3))
NUM_TRIALS = 10000 # 10 thousand
```

```python
def run_single_game(num_players=4):
  property_owners = {prop: np.random.randint(0, num_players) for prop in PROPERTIES}
  
  positions = [0] * num_players
  turn_count = 0
  current_player = 0
  
  while True:
    turn_count += 1
    
    # Roll two 6-sided dice
    dice_roll = np.random.randint(1, 7) + np.random.randint(1, 7)
    
    # Advance player and handle board wrapping (modulo 40)
    positions[current_player] = (positions[current_player] + dice_roll) % BOARD_SIZE
    current_pos = positions[current_player]
    
    # Check ownership
    if current_pos in property_owners:
      owner = property_owners[current_pos]
      if owner != current_player:
        return turn_count  # Stop game and return turn count
    
    # Move to the next player
    current_player = (current_player + 1) % num_players
```

```python
results = [run_single_game(NUM_PLAYERS) for _ in range(NUM_TRIALS)]

avg_turns = np.mean(results)
median_turns = np.median(results)

#print(f"Average turns until landing on an opponent's property: {avg_turns:.2f}")
#print(f"Median turns until landing on an opponent's property: {median_turns:.2f}")
```

```python
sns.set_theme('paper')
sns.set_style('darkgrid')
sns.set_palette(["#ff9a2a"])

# Extra stuff to save it to a file
def showplt():
  plt.figure(figsize=(10, 5))
  sns.histplot(results, bins=range(1, max(results) + 2), kde=True, color='orange', stat='probability')
  plt.axvline(avg_turns, color='red', linestyle='--', label=f'Mean: {avg_turns:.2f}')
  plt.axvline(median_turns, color='green', linestyle=':', label=f'Median: {median_turns:.2f}')
  plt.title('Distribution of Turns Until Landing on an Opponent Property')
  plt.xlabel('Number of Turns')
  plt.ylabel('Probability')
  plt.legend()
  root = tk.Tk()
  root.withdraw()
  root.attributes('-topmost', True)
  file_path = filedialog.asksaveasfilename(
    defaultextension='.png',
    filetypes=[('PNG Image', '*.png'), ('JPEG Image', '*.jpg'), ('PDF Document', '*.pdf'), ('All Files', '*.*')],
    title='Save Graph As'
  )
  if file_path:
    plt.savefig(file_path, dpi=300, bbox_inches='tight')
    print(f"Graph saved successfully to: {file_path}")
  plt.show()
  root.destroy()
```

# Conclusion

Based on the simulation of a 4-player game with properties placed every 3 spaces:

* **Expected Value:** On average, it takes **3.93 turns** (about 4 turns total) before any player lands on an opponent's property.
* **Median Value:** 50% of the simulated games resulted in a landing within **3.00 turns** or fewer.
* **Distribution Shape:** The distribution is right-skewed. While most games trigger a landing very early (within 1-3 turns), occasional long streaks without landing on an opponent's space pull the average (mean) slightly higher than the median.
